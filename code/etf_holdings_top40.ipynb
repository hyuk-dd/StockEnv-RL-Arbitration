```python
import os
import glob

import pandas as pd
```

```python
def read_csv_files(nation, market, tickers=None):
    base_path = os.path.abspath(f'../stock_data/{nation}/{market}')
    all_files = glob.glob(os.path.join(base_path, "*.csv"))

    # tickers가 주어지면 해당 ticker만 필터링
    if tickers is not None:
        tickers = set(tickers)
        all_files = [f for f in all_files if os.path.splitext(os.path.basename(f))[0] in tickers]

    ticker_list = []
    for file_path in all_files:
        ticker = os.path.splitext(os.path.basename(file_path))[0]
        ticker_list.append(ticker)

    return ticker_list
```

### 한국

```python
ks_ticker_list = read_csv_files("대한민국", "KOSPI")
ks_ticker_list = [s.removesuffix(".KS") for s in ks_ticker_list]
```

```python
df = pd.read_csv("../stock_data/etf_holdings_list/069500.KS_Holdings.csv", encoding="cp949")
df
```

```python
# 필요한 열만 남기고 비중 기준으로 내림차순 정렬
cols = ["구성종목명", "종목코드", "시가총액 구성비중"]
df = df[cols].copy()
df["시가총액 구성비중"] = pd.to_numeric(df["시가총액 구성비중"], errors="coerce")

df_ks = df.dropna(subset=["시가총액 구성비중"]).sort_values(
    "시가총액 구성비중", ascending=False).reset_index(drop=True)
# df_ks = df_ks.iloc[2:].reset_index(drop=True)
df_ks.head()
```

```python
cnt = 0
ks_top_list = []

for ticker in df_ks['종목코드']:
    if cnt == 40:
        break
    if ticker in ks_ticker_list:
        ks_top_list.append(ticker)
        cnt += 1
```

```python
len(ks_top_list)
```

```python
print(", ".join(ks_top_list))
```

### 중국
https://fundf10.eastmoney.com/ccmx_510300.html

```python
sse_ticker_list = read_csv_files("중국", "new_SSE")
sse_ticker_list = [s.removesuffix(".SS") for s in sse_ticker_list]
```

```python
df = pd.read_csv("510300.SS_Holdings.csv")

rename_map = {
    "Stock Code": "Code",
    "Stock Name": "Name",
    "Percentage of net worth": "Weight"
}
df = df.rename(columns=rename_map)

# 필요한 열만 남기고 비중 기준으로 내림차순 정렬
cols = ["Name", "Code", "Weight"]
df = df[cols].copy()
df["Code"] = pd.to_numeric(df["Code"], errors="coerce").astype("Int64").astype(str)
df["Weight"] = df["Weight"].astype(str).str.rstrip("%").astype(float)

df_510300 = df.dropna(subset=["Weight"]).reset_index(drop=True)
df_510300.head()
```

```python
cnt = 0
sse_top_list = []

for ticker in df_510300['Code']:
    if cnt == 60:
        break
    if ticker in sse_ticker_list:
        sse_top_list.append(ticker)
        cnt += 1
```

```python
len(sse_top_list)
```

```python
print(", ".join(sse_top_list))
```
